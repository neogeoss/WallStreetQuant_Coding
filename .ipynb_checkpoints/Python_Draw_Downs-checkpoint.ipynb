```python
import yfinance as yf
import pandas as pd
%matplotlib inline
```

```python
univ = ['TSLA','SPY']
px = yf.download(univ, start='2021-01-01')['Close']
```

```python
px
```

```python
def drawdown(px):
    return (px/px.expanding(min_periods=1).max() - 1)
```

```python
dd = drawdown(px)
```

```python
dd.plot()
```

```python
dd.min()
```

```python
def duration(px):
    
    peak = px.expanding(min_periods=1).max()

    res = pd.DataFrame(index=px.index, columns = px.columns)

    for col in px.columns:
        for dt in px.index:
            if px.loc[dt,col] >= peak.loc[dt, col]:
                res.loc[dt, col]=0

            elif dt == res.index[0]:
                res.loc[dt, col] = 1

            else:
                res.loc[dt,col] = res.loc[:dt, col].iloc[-2] + 1

    return res
```

```python
ddd = duration(px)
```

```python
ddd.plot()
```

```python
ddd.max()
```

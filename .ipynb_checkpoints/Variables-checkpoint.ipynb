```python
tickers = ['AAPL US Equity', 'WMT US Equity', '1U1 GR Equity']

us_tickers = [x.split(" ")[0] for x in tickers if 'US Equity' in x]
print(us_tickers)
```

```python
## Create a dictionary in one line
ctry_map = {x.split(" ")[0]: x.split(' ')[1] for x in tickers}
print(ctry_map)
```
